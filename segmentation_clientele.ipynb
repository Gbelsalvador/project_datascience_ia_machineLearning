**La segmentation client** consiste à regrouper les clients en fonction de caractéristiques, comportements et préférences partagés. En segmentant les clients, les entreprises peuvent adapter leurs stratégies et cibler des groupes spécifiques plus efficacement et augmenter la valeur globale du marché.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.cluster import KMeans
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("data/new.csv")
df.head()
```

```python
for col in df.columns:
    temp = df[col].isnull().sum()
    if temp > 0:
        print(f'Column {col} contains {temp} null values.')
```

```python
df = df.dropna()
print("Total values in the dataset after removing the null values:", len(df))
```

```python
df.nunique()
```

```python
parts = df["Dt_Customer"].str.split("-", n=3, expand=True)
df["day"] = parts[0].astype('int')
df["month"] = parts[1].astype('int')
df["year"] = parts[2].astype('int')
```

```python
df.drop(['Z_CostContact', 'Z_Revenue', 'Dt_Customer'], axis=1, inplace=True)
```

```python
# 1. Sélectionner les colonnes de campagnes
accepted_cols = [col for col in df.columns if 'AcceptedCmp' in col or 'Accepted' in col]

# 2. Trouver la première colonne où la valeur vaut 1 pour chaque ligne
# .idxmax(axis=1) renvoie le nom de la colonne qui contient la valeur max (1)
df['Accepted'] = df[accepted_cols].idxmax(axis=1)

# Si le client n'a accepté aucune campagne (que des 0), idxmax mettra quand même la 1ère colonne.
# Pour corriger et mettre "None" ou "Aucune" quand la somme est 0 :
df.loc[df[accepted_cols].sum(axis=1) == 0, 'Accepted'] = 'Aucune'
```

```python
objects = df.select_dtypes(include=['object']).columns.tolist()
floats = df.select_dtypes(include=['float']).columns.tolist()

print(objects)
print(floats)
```

```python
plt.subplots(figsize=(15, 10))
for i, col in enumerate(objects):
    plt.subplot(2, 2, i + 1)
    sns.countplot(df[col])
plt.show()
```

```python
plt.subplots(figsize=(15, 10))
for i, col in enumerate(objects):
    plt.subplot(2, 2, i + 1)

    df_melted = df.melt(id_vars=[col], value_vars=['Response'], var_name='hue')
    sns.countplot(x=col, hue='value', data=df_melted)
plt.show()
```

```python
object_cols = df.select_dtypes(include=['object']).columns
le = LabelEncoder()
for col in object_cols:
    df[col] = le.fit_transform(df[col])
```

```python
plt.figure(figsize=(15, 15))
sns.heatmap(df.corr() > 0.8, annot=True, cbar=False)
plt.show()
```

```python
scaler = StandardScaler()
data = scaler.fit_transform(df)
```

```python
from sklearn.manifold import TSNE
model = TSNE(n_components=2, random_state=0)
tsne_data = model.fit_transform(df)
plt.figure(figsize=(7, 7))
plt.scatter(tsne_data[:, 0], tsne_data[:, 1])
plt.show()
```

```python
error = []
for n_clusters in range(1, 21):
    model = KMeans(init='k-means++',
                   n_clusters=n_clusters,
                   max_iter=500,
                   random_state=22)
    model.fit(df)
    error.append(model.inertia_)
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(x=range(1, 21), y=error)
sns.scatterplot(x=range(1, 21), y=error)
plt.show()
```

```python
model = KMeans(init='k-means++',
               n_clusters=5,
               max_iter=500,
               random_state=22)
segments = model.fit_predict(df)
```

```python
plt.figure(figsize=(7, 7))

df_tsne = pd.DataFrame({'x': tsne_data[:, 0], 'y': tsne_data[:, 1], 'segment': segments})

sns.scatterplot(x='x', y='y', hue='segment', data=df_tsne)
plt.show()
```
